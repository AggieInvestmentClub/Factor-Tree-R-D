# What the short side adds

**Design notes - 2026-10-29**

Proposed additions to the existing v3.1 payoff + ETF-context long/cash study. 

The aim is to test whether selective short exposure improves net Sharpe after costs and improves max drawdown for each of the same 15 ETFs, and provides us enough conviction to continue with developing a put option side to this model.

## 1. A third state: SHORT

| Existing decision | Proposed decision | Meaning |
|---|---|---|
| LONG | LONG | Own positive ETF exposure when the estimated net long opportunity qualifies. |
| CASH / FLAT | FLAT | Hold no ETF exposure when neither direction qualifies, estimates tie, or calibration support is insufficient. |
| No short exposure | SHORT | Borrow and sell ETF exposure when the estimated net short opportunity qualifies and shorting is eligible. Later buy it back to cover. |

The new short state is selective. A rejected long can still have a weak positive expected return, an uncertain forecast, or a decline too small to cover shorting costs. Therefore, flat does not automatically become short.

When borrowing is unavailable, only the short candidate becomes ineligible. A separately qualifying long can still be selected.

## 2. Separate estimates of long and short net payoff

The first extension can reuse the existing context-tree forecasts while estimating expected net long payoff and expected net short payoff separately from prior matured out-of-fold observations. Out-of-fold means that each historical prediction was generated without fitting on its own outcome.

This adds a directional payoff assessment, rather than requiring a new three-class tree or an independently trained short tree immediately. Those could be separate later hypotheses.

Long and short estimates each deduct their own costs. Simply negating an already-net long outcome incorrectly turns its deducted costs into a credit. Likewise, one minus the probability of a profitable long is not automatically the probability of a profitable short.

Short outcomes need a consistent treatment of borrowing and distributions. Adjusted total-return prices can be a research proxy, but separate dividend debits must not double-count distributions already represented by that proxy. The accounting convention needs to be declared before a run.

## 3. A neutral region between the directions

Each eligible direction must clear its own nonnegative net-payoff threshold. Choose the best qualifying direction; if neither qualifies, stay flat. A tie or insufficient calibration support also goes flat.

| Situation | Proposed state |
|---|---|
| Long is the best eligible opportunity above its threshold. | LONG |
| Short is eligible and has the best opportunity above its threshold. | SHORT |
| Neither direction clears its threshold. | FLAT |
| Eligible estimates tie, or calibration support is insufficient. | FLAT |
| Short is unavailable but long independently qualifies. | LONG |

Costs already create a region in which trading is unattractive; an additional threshold can widen that neutral region. Numeric thresholds and minimum support remain open decisions. They must be declared using earlier chronological research before evaluation, rather than chosen from the evaluated interval's results.

## 4. Signed exposure and short-aware account records

The state supplies the sign, while the existing fixed or past-volatility sizing convention supplies the magnitude. Proposed target exposure ranges from -100% to +100% of account NAV: positive for long, zero for flat and negative for short.

The existing completed-weekly-close decision and next-actual-session-open execution clock remain. Forecast horizon and actual holding duration remain different. Moving from long to short closes one position and opens the other, so actual turnover includes both legs.


## 5. Short-specific costs, data and risk rules

| Addition | Why it matters |
|---|---|
| Historical borrow availability | A negative forecast cannot establish that shares could have been borrowed. |
| Borrow charges | Fees reduce short payoff and need a declared accrual convention, including weekends and calendar-day gaps. |
| Distributions | The lender's distribution entitlement must be represented consistently with the chosen price/accounting convention. |
| Margin and collateral | Short exposure has obligations even when the target magnitude is no more than account NAV. |
| Recalls and forced covers | A position may need to close for reasons other than a forecast change. |
| Terminal cover | Remaining shorts must be closed at the declared end of the simulation, with costs. |
| Gaps and insolvency | Failed or insolvent account paths cannot be silently omitted from reported results. |


## 6. The planned comparison

For each ETF, the next study would compare three separate procedures on matched dates:

1. The proposed long / flat / short extension.
2. The frozen long / cash payoff + ETF-context control.
3. The ETF's buy-and-hold account.

Net daily Sharpe after costs remains the primary measure, however equity and drawdown metrics are very important as well and will determine pure profitability. Reports should also separate long and short PnL, wins, holding duration, signed versus absolute exposure, turnover, transaction costs and borrow drag. Signed exposure alone can conceal how much risk was taken when long and short periods offset in the average.

Borrow/fee sensitivity and paired uncertainty should accompany any historical result. A useful question is whether shorting adds an edge after costs, rather than merely changing exposure or benefiting in one selected period. Each of the 15 accounts is evaluated independently; these comparisons do not construct a portfolio.


## 7. What stays fixed, and what still needs a decision

For the first controlled extension, the source study's model procedure will be preserved with ETF-context inputs, year-specific horizons, sizing conventions, original forecasts and published long-only results.

The main unresolved choices are the directional thresholds and calibration support, short-availability assumptions, borrowing scenarios, distribution convention, and margin/recall/cover handling. These choices need explicit definitions before the future backtest.

The same 15 ETFs remain in scope: XLK, XLF, XLE, XLV, XLI, XLY, XLP, XLU, XLB, IEF, TLT, TIP, GLD, DBC and VNQ.
